```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('Iris Dataset.csv')
df.head()
```

```python
sns.barplot(x="Species",y="SepalLengthCm",data=df,palette='rainbow')
plt.title('Species vs SepalLengthCm')
plt.xlabel('Species')
plt.ylabel('SepalLengthCm')
plt.show()
```

```python
sns.countplot(data=df, x="Species")
plt.show()
```

```python
sns.boxplot(x="Species",y="SepalWidthCm",data=df, palette='turbo')
plt.title('Species vs SepalWidthCm')
plt.xlabel('Species')
plt.ylabel('SepalWidthCm')
plt.show()
```

```python
sns.swarmplot(x="Species",y="SepalWidthCm",data=df, palette='turbo')
plt.title('Species vs SepalWidthCm')
plt.xlabel('Species')
plt.ylabel('SepalWidthCm')
plt.show()
```

```python
sns.distplot(df['SepalWidthCm'])
```

```python
sns.jointplot(df['PetalWidthCm'])
```

```python
sns.pairplot(data=df,hue='Species',palette='rainbow')
```
